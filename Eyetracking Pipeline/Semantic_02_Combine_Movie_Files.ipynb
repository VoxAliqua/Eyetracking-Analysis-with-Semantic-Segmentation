```python
import pandas as pd
import os

# Define the root directory path
root_directory = (r"E:\University\Master\4.SS22\IMaC Lab\Shamem\EyeTracking\Data\Semantic_01_Segmentation\Robust_batch")

output_directory = (r"E:\University\Master\4.SS22\IMaC Lab\Shamem\EyeTracking\Data\Semantic_01_Segmentation\Robust_batch")

# Get a list of all subdirectories in the root directory
subdirectories = [subdir for subdir in os.listdir(root_directory) if os.path.isdir(os.path.join(root_directory, subdir))]


# Iterate over the subdirectories
for subdir in subdirectories:
    # Create a directory path for the current subdirectory
    subdir_path = os.path.join(root_directory, subdir)
    
    # Get a list of CSV files in the current subdirectory
    csv_files = [file for file in os.listdir(subdir_path) if file.endswith('.csv')]
    
    # Create an empty DataFrame to store the combined frames
    combined_frames = pd.DataFrame()
    
    # Iterate over the CSV files
    for csv_file in csv_files:
        # Load each CSV file into a DataFrame
        frame_data = pd.read_csv(os.path.join(subdir_path, csv_file), header=None)
        
        # Split the pixel values into separate columns
        split_frames = frame_data[0].str.split(';', expand=True)
        
        # Add a new column for the frame number
        frame_number = int(csv_file.split('.')[0].split('_')[-1])
        split_frames.insert(0, 'Frame', frame_number)
        
        # Append the split frames to the combined frames DataFrame
        combined_frames = pd.concat([combined_frames, split_frames], ignore_index=True)
    
        
    # Define the output filename based on the subdirectory name
    output_filename = f'{subdir}_combined.csv'
    
    # Save the combined frames to a new CSV file without the index
    output_path = os.path.join(output_directory, output_filename)
    combined_frames.to_csv(output_path, index=False)
```
